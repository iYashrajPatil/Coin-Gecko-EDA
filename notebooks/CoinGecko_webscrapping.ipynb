import requests
import pandas as pd

def scrape_coingecko():
    url = "https://api.coingecko.com/api/v3/coins/markets"
    params = {
        "vs_currency": "usd",
        "order": "market_cap_desc",
        "per_page": 100,  # number of coins per page
        "page": 1,
        "sparkline": False
    }

    response = requests.get(url, params=params, timeout=10)
    data = response.json()

    # convert to DataFrame
    df = pd.DataFrame(data, columns=[
        "id", "symbol", "name", "current_price", "market_cap",
        "total_volume", "high_24h", "low_24h", "price_change_percentage_24h"
    ])

    df.to_csv("coingecko_top100.csv", index=False)
    return df

df = scrape_coingecko()

```python
import requests
import pandas as pd
import time

def scrape_all_coins():
    url = "https://api.coingecko.com/api/v3/coins/markets"
    all_data = []
    page = 1

    while True:
        params = {
            "vs_currency": "usd",
            "order": "market_cap_desc",
            "per_page": 250,  # max allowed
            "page": page,
            "sparkline": False
        }

        response = requests.get(url, params=params, timeout=30)

        if response.status_code == 429:  # too many requests
            time.sleep(60)
            continue

        if response.status_code != 200:
            print(f"Error: {response.status_code}")
            break

        data = response.json()
        if not data:
            break

        all_data.extend(data)
        page += 1

        time.sleep(6)  #  wait between requests (10/min)

    # Convert to DataFrame
    df = pd.DataFrame(all_data, columns=[
        "id", "symbol", "name", "current_price", "market_cap",
        "total_volume", "high_24h", "low_24h", "price_change_percentage_24h"
    ])

    df.to_csv("coingecko_all_coins.csv", index=False)
    return df

df = scrape_all_coins()
print(f"\nScraped {len(df)} coins!")
print(df.head())
```
